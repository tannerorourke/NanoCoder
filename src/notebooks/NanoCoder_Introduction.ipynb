### About NanoCoder

NanoCoder is a case study where we build, train, and fine-tune a small yet powerful LLAMA-style language model specializing in technical comprehension and Python code generation.

In the coming series of notebooks, you will:

- Create a Hugging Face dataset from **7** sources
- Train it on an interleaved, code-heavy corpus, curated to promote specialization in technical explanation and Pythonic coding, the same corpus we will later train on.
- Design and implement a byte-level "Semi-Supervised" BPE tokenizer from scratch
- Build a GPT-2 style Language Model
- Upgrade that LM with RoPE, FIM, SwiGLU, and RMSnorm
- Perform supervised fine-tuning (SFT), Rejection Fine-tuning (RFT), and Direct Preference Optimization (DPO)
- Export each to HuggingFace (Datasets, Tokenizers, Models)

The complete datasets (pretrain, SFT, and RLHF), base model, and instruct models (at each step) are available on [my HuggingFace](https://huggingface.co/torq1).

```python
import torch
from huggingface_hub import snapshot_download
from nanocoder.model.nanocoder import NanoCoder

# Pull the finished instruct model - the one this whole series builds up to - and load it.
# from_pretrained restores the weights, config, and tokenizer together from one folder.
path = snapshot_download("torq1/NanoCoder-123M-sft")
device = "cuda" if torch.cuda.is_available() else "cpu"
nano = NanoCoder.from_pretrained(path, device=device)

# The model answers in a '## Task / ## Solution' shape - the prompt interface we design in NanoCoder 1.
prompt = "## Task\nWrite a function that adds two numbers.\n\n## Solution\n"
print(nano.generate_text(prompt, max_new_tokens=256, temperature=0.7, top_k=20))
```
